### 로그 변환 적용 및 새로운 모델 구축

선형성 가정 개선을 위해 종속변수 `visitors`와 주요 연속형 독립변수들에 `np.log1p`를 적용하여 로그 변환된 변수들을 생성합니다. `np.log1p(x)`는 `log(1+x)`를 계산하므로, 0을 포함한 값에도 안전하게 로그 변환을 적용할 수 있습니다.

```python
import numpy as np
import statsmodels.api as sm

# 로그 변환된 변수 생성 (0 값을 포함할 수 있는 변수에 np.log1p 사용)
df['log_visitors'] = np.log1p(df['visitors'])
df['log_admission_fee($)'] = np.log1p(df['admission_fee($)'])
df['log_culture_budget($M)'] = np.log1p(df['culture budget($M)'])
df['log_per_capita_GDP($)'] = np.log1p(df['per capita GDP($)'])
df['log_num_overseas_tourist'] = np.log1p(df['Number of overseas tourist (in millions)'])

# 새로운 종속변수와 설명변수 목록
y_log = df['log_visitors']

X_vars_log = [
    'log_admission_fee($)',
    'log_culture_budget($M)',
    'log_per_capita_GDP($)',
    'log_num_overseas_tourist',
    'COVID_Dummy',
    'Disaster_Dummy',
    'Social_Unrest',
    'MegaEvent_Dummy'
]

X_log = sm.add_constant(df[X_vars_log])

# 로그 변환된 변수로 새로운 OLS 모델 적합
model_log = sm.OLS(y_log, X_log).fit()

print(model_log.summary())
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import statsmodels.api as sm

# --- (이미 만든 model_multi를 사용) ---
# model_multi = sm.OLS(y, X).fit()

def ols_diagnostic_plots(model, X, y, top_n_influential=5):
    """
    model: statsmodels OLSResults (e.g., model_multi)
    X: design matrix used in fit (including constant)
    y: dependent variable series/array
    """

    fitted = model.fittedvalues
    resid = model.resid

    # Standardized (studentized) residuals
    infl = model.get_influence()
    resid_std = infl.resid_studentized_internal
    leverage = infl.hat_matrix_diag
    cooks = infl.cooks_distance[0]

    # 1) Actual vs Predicted
    plt.figure(figsize=(7,5))
    plt.scatter(fitted, y, alpha=0.7)
    # 45-degree line
    y_min, y_max = np.min(y), np.max(y)
    plt.plot([y_min, y_max], [y_min, y_max])
    plt.xlabel("Predicted (fitted) values")
    plt.ylabel("Actual values (y)")
    plt.title("Actual vs Predicted (OLS)")
    plt.tight_layout()
    plt.show()

    # 2) Residuals vs Fitted (linearity / heteroskedasticity)
    plt.figure(figsize=(7,5))
    plt.scatter(fitted, resid, alpha=0.7)
    plt.axhline(0, linestyle="--")
    plt.xlabel("Fitted values")
    plt.ylabel("Residuals")
    plt.title("Residuals vs Fitted")
    plt.tight_layout()
    plt.show()

    # 3) Q-Q plot (normality of residuals)
    plt.figure(figsize=(7,5))
    sm.qqplot(resid, line='45', fit=True)
    plt.title("Normal Q-Q plot of residuals")
    plt.tight_layout()
    plt.show()

    # 4) Scale-Location plot (sqrt(|standardized residuals|) vs fitted)
    plt.figure(figsize=(7,5))
    plt.scatter(fitted, np.sqrt(np.abs(resid_std)), alpha=0.7)
    plt.xlabel("Fitted values")
    plt.ylabel("Sqrt(|Standardized residuals|)")
    plt.title("Scale-Location (Homoskedasticity check)")
    plt.tight_layout()
    plt.show()

    # 5) Influence: Cook's distance
    plt.figure(figsize=(7,5))
    plt.stem(np.arange(len(cooks)), cooks, markerfmt=" ", basefmt=" ")
    plt.xlabel("Observation index")
    plt.ylabel("Cook's distance")
    plt.title("Influence (Cook's distance)")
    plt.tight_layout()
    plt.show()

    # label top influential points
    top_idx = np.argsort(cooks)[-top_n_influential:]
    print(f"Top {top_n_influential} influential points (by Cook's distance):")
    print(top_idx, cooks[top_idx])

    # 6) Coefficient plot with 95% CI (excluding constant)
    params = model.params.drop(labels=["const"], errors="ignore")
    conf = model.conf_int().loc[params.index]
    coef = params.values
    lower = conf[0].values
    upper = conf[1].values

    plt.figure(figsize=(8, max(4, 0.4*len(params))))
    y_pos = np.arange(len(params))
    plt.errorbar(coef, y_pos, xerr=[coef - lower, upper - coef], fmt='o')
    plt.axvline(0, linestyle="--")
    plt.yticks(y_pos, params.index)
    plt.xlabel("Coefficient (with 95% CI)")
    plt.title("Coefficient Plot (OLS)")
    plt.tight_layout()
    plt.show()


# 실행
ols_diagnostic_plots(model_multi, X, y, top_n_influential=5)
```
